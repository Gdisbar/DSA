## 15. 3Sum

Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.
Notice that the solution set must not contain duplicate triplets.
```bash

Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
Explanation: 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.
The distinct triplets are [-1,0,1] and [-1,-1,2].
Notice that the order of the output and the order of the triplets does not matter.

Input: nums = [0,1,1]
Output: []
Explanation: The only possible triplet does not sum up to 0.

Input: nums = [0,0,0]
Output: [[0,0,0]]
Explanation: The only possible triplet sums up to 0.
```

```python
"""
twoSum(i=k+1,j=n-1,target=-1*nums[k]) 

k + 1: You start searching for the other two numbers from the element immediately following k. This avoids reusing the same element and helps avoid duplicate triplets.

nums = [-4, -1, 1, 3, 5]
# Indices: 0,  1, 2, 3, 4

When k = 0 (nums[k] = -4):
You search the remaining elements from index 1 to 4 ([-1, 1, 3, 5]). You find the valid triplet: [-4, 1, 3].

When k = 2 (nums[k] = 1):
If you didn't start at k + 1 and instead looked backward or started at k, you might look at index 0 (-4) again. You would find [1, -4, 3], which is the exact same triplet [-4, 1, 3] you already found!

By enforcing i = k + 1, you guarantee an order (k < i < j), which perfectly cuts out redundant paths and guarantees that every triplet found is unique.
"""

def threeSum(self, nums: List[int]) -> List[List[int]]:
        n = len(nums)
        if n < 3:
            return []
        
        def twoSum(i,j,target):
            while i < j:
                if nums[i]+nums[j] > target:
                    j-=1
                elif nums[i]+nums[j] < target:
                    i+=1
                else:
                    while i+1< n and i < j and nums[i]==nums[i+1]:
                        i+=1
                    while j-1 > 0 and i < j and nums[j]==nums[j-1]:
                        j-=1
                    result.append([nums[i],nums[j],-1*target])
                    i+=1
                    j-=1
        result = []
        nums = sorted(nums)
        for k in range(n-2):
            if k!=0 and nums[k]==nums[k-1]:
                continue
            twoSum(k+1,n-1,-1*nums[k])
        return result

```


## 16. 3Sum Closest

Almost same as 3Sum instead of appending the triplet to result - check it's closest to given target value and append the sum i.e abs(sum_2-target)

No need for duplicate checking

```python
"""
1. we update at 
if abs(sum_2-target) < abs(target-closestSum):
    closestSum = sum_2

2. No need for duplicate check
3. sum_2 = arr[i]+arr[j]+arr[k] # here i = k+1, since we're sending k not k+1

for k in range(n-2):
    twoSum(k,target)
"""

def threeSumClosest(self, nums: list[int], target: int) -> int:
    nums.sort()
    closest_sum = float('inf')
    n = len(nums)
    
    for i in range(n - 2):
        left = i + 1
        right = n - 1
        
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]
            
            # If current_sum is closer to target than our 
            # previous best, update it
            if abs(target - current_sum) < abs(target - closest_sum):
                closest_sum = current_sum
            # Move pointers based on comparison with target
            if current_sum > target:
                right -= 1
            elif current_sum < target:
                left += 1
            else:
                # it's the closest possible
                return current_sum
                
    return int(closest_sum)
```

## Factorials of large numbers

```python

def factorial(self, n):
    # storing factorial upto 1000 safely
    nums = [0 for _ in range(3000)] 
    nums[0]=1
    res_size = 1
    def multiply(x:int):
        # nonlocal - variable defined in an outer (parent) function
        # global - variable defined in top level of a file/module
        nonlocal res_size 
        carry = 0
        for i in range(res_size):
            result = x*nums[i]+carry
            nums[i]=result%10
            carry = result//10
        while carry>0:
            nums[res_size]=carry%10
            carry=carry//10
            res_size+=1

    for i in range(2,n+1):
        multiply(i)
    
    result = []
    for i in range(res_size-1,-1,-1):
        result.append(nums[i])
    return result

```

# Traffic

You're given a number "N", "M" and an array of vehicles cosisting of 
0's & 1's of length "N", You can perform at most "M" flips where one 
flip can change '0' -> '1'. Print out the longest sequence of '1's you 
can make using "M" moves

```bash
Input :
6 3
0 1 0 0 1 0
Output:
5 [0,1,1,1,1,1]

```

```python
"""
zero_count : This keeps track of how many empty spots (or "flips") you are currently using within your window.

If your zero_count exceeds m, your window has become "invalid" (you are trying to flip more spots than allowed).

To fix this, you move the left pointer forward. As you move left past a 0, you decrement zero_count until the window is valid again (zero_count <= m).

"""


def traffic(n: int, m: int, vehicle: [int]) -> int:
    left,right = 0,0
    zero_count = 0
    max_window = 0
    while right < n:
        if vehicle[right]==0:
           zero_count +=1
        while zero_count > m:
            if vehicle[left]==0:
                zero_count -=1
            left +=1
        max_window = max(max_window,right-left+1)
        right+=1
        
    return max_window

```

## 31. Next Permutation

The next permutation of an array of integers is the next 
lexicographically greater permutation of its integer. 

- For example, the next permutation of arr = [1,2,3] is [1,3,2].
- Similarly, the next permutation of arr = [2,3,1] is [3,1,2].
- While the next permutation of arr = [3,2,1] is [1,2,3] because [3,2,1] does not have a lexicographical larger rearrangement.

The replacement must be in place and use only constant extra memory.

```bash

nums = [1, 5, 8, 4, 2]
Find Pivot: Scanning from the right, 4 <= 8 (True), 2 <= 4 (True), but 8 > 5 (True). The pivot is a[1] which is 5.

Find Successor: Scan from the right for the first number greater than 5. That is 8.

Swap: Swap 5 and 8. The list becomes [1, 8, 5, 4, 2].

Reverse Tail: Reverse the part after index 1 (the [5, 4, 2] part). It becomes [2, 4, 5].

Final Result: [1, 8, 2, 4, 5].
```

```python
"""
1. Find the "Pivot" : scan from right to left to find the first element that is smaller than the one to its right. This index i is the "pivot." If the entire array is in descending order (e.g., [3, 2, 1]), the loop will finish with i = -1, meaning it's already the largest possible permutation.

2. Find the "Successor": If a pivot was found, you look from right to left again to find the first number that is greater than a[i]. You swap them. This swap ensures that the number at i is now slightly larger, which is the necessary step to create a "next" greater permutation.

3. Reverse the Tail: After the swap, the elements to the right of the original pivot i are still in descending order. To make the permutation as small as possible (the next one), you need to reverse that tail into ascending order.

"""


def nextPermutation(self, nums: List[int]) -> None:
    """
    Do not return anything, modify nums in-place instead.
    """
    a = nums
    n = len(a)
    # Find the "Pivot"
    i = n - 2
    while i >=0 and a[i+1] <= a[i]:
        i -= 1
    # Find the "Successor"
    if i >= 0:
        j = n -1
        while j>=0 and a[j] <= a[i]:
            j -= 1
        a[i],a[j] = a[j],a[i]
    # Reverse the Tail
    i = i + 1
    j = n -1
    while i < j:
        a[i],a[j] = a[j],a[i]
        i += 1
        j -= 1
    return a
```

## 48. Rotate Image

You are given an n x n 2D matrix representing an image, rotate the image by 90 degrees (clockwise).

You have to rotate the image in-place, which means you have to modify the input 2D matrix directly. DO NOT allocate another 2D matrix and do the rotation.

```bash
Input: matrix = [
                [1,2,3],
                [4,5,6],
                [7,8,9]
            ]
Output: [
        [7,4,1],
        [8,5,2],
        [9,6,3]
    ]
```

```python

def rotate(self, matrix: list[list[int]]) -> None:
    """
    Do not return anything, modify matrix in-place instead.
    """
    # 1. Reverse the matrix (rows order)
    matrix.reverse()
    
    # 2. Transpose the matrix (swap matrix[i][j] with matrix[j][i])
    n = len(matrix)
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
```

## 11. Container With Most Water


You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) and (i, height[i]).

Find two lines that together with the x-axis form a container, such that the container contains the most water.

Return the maximum amount of water a container can store.

Notice that you may not slant the containe

```bash
Input: height = [1,8,6,2,5,4,8,3,7]
Output: 49
Explanation: The above vertical lines are represented by array [1,8,6,2,5,4,8,3,7]. In this case, the max area of water (blue section) the container can contain is 49.
```

```python
"""
Maximize height - Classic Two pointer

height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
Start at l=0 (height 1) and r=8 (height 7).
Width is 8, height is min(1, 7) = 1. Area = 8 * 1 = 8.
Since the left side (1) is shorter, move l to the right (l=1).

Now l=1 (height 8) and r=8 (height 7).
Width is 7, height is min(8, 7) = 7. Area = 7 * 7 = 49.
The right side (7) is now shorter, so move r to the left.

Keep repeating this until the pointers meet.

"""
def maxArea(self, height: list[int]) -> int:
    l = 0
    r = len(height) - 1
    area = 0
    while l < r:
        h = min(height[l], height[r])
        width = r - l
        area = max(area, h * width)
        # Maximize height
        if height[l] <= height[r]:
            l += 1
        else:
            r -= 1
    return area
```
