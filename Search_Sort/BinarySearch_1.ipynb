## 153. Find Minimum in Rotated Sorted Array

Given the sorted rotated array nums of **unique elements**, return the minimum element of this array.
```bash
Input: nums = [4,5,6,7,0,1,2]
Output: 0
Explanation: The original array was [0,1,2,4,5,6,7] and it was rotated 4 times.
```
```python
"""
Case: nums[mid] <= nums[r]
---------------------------------------------------
This means the segment from mid to r is sorted — the minimum can’t be further right, but might be mid itself.

Example: [4, 5, 6, 7, 0, 1, 2]. If mid points to 7 and r points to 2, 7 <= 2 is false. But if mid points to 0 and r points to 2, 0 <= 2 is true.

If nums[mid] <= nums[r], the minimum must be at mid or to its left. So, we set r = mid.

Case: nums[mid] > nums[r]
-----------------------------
Rotation happened after mid,smallest element is not in the left part (including mid).

Example: In [4, 5, 6, 7, 0, 1, 2], if mid points to 5 and r points to 2, 5 > 2 is true. The pivot must be to the right of mid.

We set l = mid + 1.

since mid is a potential answer we can't use l <= r  , 
since in that case when l == r , mid will oscillate between l & r
result is stored in both l,r
"""

def findPivot(self,nums:List[int]) -> int:
    l , r = 0, len(nums)-1
    while l < r:
        # no duplicate check
        mid = l + (r-l)//2
        if nums[mid]<= nums[r]:
            r = mid
        else:
            l = mid+1
    return nums[l] # return pivot element not index

def findMin(self, nums: List[int]) -> int:
    return self.findPivot(nums)
        
```

## 81. Search in Rotated Sorted Array II

There is an integer array nums sorted in non-decreasing order (**not necessarily with distinct values**).

Given the array nums after the rotation and an integer target, return true if target is in nums, or false if it is not in nums.
```bash
Input: nums = [2,5,6,0,0,1,2], target = 0
Output: true

Input: nums = [2,5,6,0,0,1,2], target = 3
Output: false
```

```python
# Approach-1

def findPivot(self,nums:List[int]) -> int:
    l , r = 0, len(nums)-1
    while l < r:
        # no duplicate check
        mid = l + (r-l)//2
        if nums[mid]<= nums[r]:
            r = mid
        else:
            l = mid+1
    return l # return pivot index

def binarySearch(self,nums: List[int],l:int,r:int,target:int):
    while l<=r:
        m = l + (r-l)//2
        if nums[m]==target:
            return m
        elif nums[m] > target:
            r-=1
        else:
            l+=1
    return -1
def search(self, nums: List[int], target: int) -> int:
    n = len(nums)
    pivot = self.findPivot(nums)
    if nums[pivot]==target:
        return pivot

    idx = self.binarySearch(nums,0,pivot-1,target)
    if idx !=-1:
        return idx
    idx = self.binarySearch(nums,pivot+1,n-1,target)
    return idx

# Approach -2

"""
After rotation [0, 1, 2, 4, 5, 6, 7] becomes [4, 5, 6, 7, 0, 1, 2]. 
However, the beauty of this array is that at least one half (either the 
left side or the right side of m) is always sorted.

1. Find the middle - If nums[m] == target, you found it!
2. Identify the sorted half 
If nums[l] <= nums[m]: The left side is sorted. 
 - Now, check if the target exists within this sorted range 
   (nums[l] <= target < nums[m]).
 - If yes, search the left side: h = m - 1.
 - If no, the target must be in the "unsorted" right side: l = m + 1.

Else (nums[m] < nums[l]): The right side is sorted.
 - Check if the target exists within this sorted range 
   (nums[m] < target <= nums[h]).
 - If yes, search the right side: l = m + 1.
 - If no, the target must be in the "unsorted" left side: h = m - 1.

---------------------------
[4, 5, 6, 7, 0, 1, 2], target = 0
l=0, h=6, m=3 (value 7).
nums[l] (4) is <= nums[m] (7). The left side [4, 5, 6, 7] is sorted.
Is target (0) between 4 and 7? No.
Move l to m + 1 (l=4).
New range: [0, 1, 2].
The logic continues until nums[m] matches target.
"""
def search(self, nums: List[int], target: int) -> int:
    l = 0
    h = len(nums) - 1
    while l <= h:
        m = (l+h)//2
        if nums[m]==target:
            return m
        elif nums[l] <= nums[m]:
            if target > nums[m] or target < nums[l]:
                l = m + 1
            else:
                h = m - 1
        else:
            if target < nums[m] or nums[h] < target:
                h = m -1
            else:
                l = m + 1
    return -1


```

## 74. Search a 2D Matrix

Given an integer target, return true if target is in matrix or false otherwise.

- Each row is sorted in non-decreasing order.
- The first integer of each row is greater than the last integer of the previous row.
```bash
matrix = [[1,3,5,7],
                [10,11,16,20],
                [23,30,34,60]
                ], target = 3
Output: true
```


```python
"""
Approach-1 O(m*n)
---------------------------------
i = 0, j = n-1

1. matrix[i][j] > target -> target in same row but previous column # search space left
2. matrix[i][j] < target -> target in next row # search space right

Approach-2 (Using Binary Search) O(log(m*n))
--------------------------------------------------
start = 0, end   = m*n-1 ## flatten matrix
1. mapping 2D matrix to 1D flatten array
   row = mid/n;
   col = mid%n;
2. matrix[row][col] > target -> end will decrease # search space left,move to same row prev. column
3. matrix[row][col] < target -> start will increase # search space right, move to next row
"""

while(start <= end) {
   int mid = start + (end-start)//2;
   
   int row = mid/n;
   int col = mid%n;
   
   if(matrix[row][col] > target) {
         end = mid-1;
   } else if(matrix[row][col] < target) {
         start = mid+1;
   } else {
         return true;
   }
}
```

## 540. Single Element in a Sorted Array

You are given a sorted array consisting of only integers where every element appears exactly twice, except for one element which appears exactly once.

Return the single element that appears only once.

    Input: nums = [1,1,2,3,3,4,4,8,8]
    Output: 2

```python
"""
Approach : 1 (Using XOR operator) : O(n)
-----------------------------------------------
1 ^ 7 = 1, 6 ^ 6 = 0
xor_Value = 0
for each element -> xor_Value ^= a[i]

Approach : 2 (Using Binary Search) : O(log(n))
---------------------------------------------------
1. The Pattern - Before the single element, pairs appear at (even, odd) 
indices. After the single element, the pattern shifts to (odd, even) 
indices.

Example: [1, 1, 2, 3, 3]
Index 0, 1: (1, 1) — Even index 0, Odd index 1.
Index 2: 2 (The shift happens here!)
Index 3, 4: (3, 3) — Odd index 3, Even index 4.

2. The Binary Search Strategy
The code uses binary search to find the point where this pattern breaks. Instead of just checking if nums[mid] equals its neighbor, it checks the parity of the remaining elements (is_even).

If nums[mid] == nums[mid + 1]: This is a "normal" pair.

- If the number of elements to the right (r - mid) is even, 
  it means the single element is further to the right.
- If odd, the single element is to the left.

If nums[mid] == nums[mid - 1]: This is a pair shifted left.

- If is_even is true, the single element is to the left.
- Otherwise, it's to the right.

The "Found" Case: If nums[mid] is not equal to either neighbor, you have landed directly on the unique element.

"""

def singleNonDuplicate(self, nums: list[int]) -> int:
    n = len(nums)
    l, r = 0, n - 1
    
    while l < r:
        mid = l + (r - l) // 2
        # Check if the number of elements from mid to r is even
        is_even = (r - mid) % 2 == 0
        
        if nums[mid] == nums[mid + 1]:
            # If the pair is (mid, mid+1), we check the right side
            if is_even:
                # If the remaining elements are even, the single element 
                # must be to the right of this pair.
                l = mid + 2
            else:
                r = mid - 1
        elif nums[mid] == nums[mid - 1]:
            # If the pair is (mid-1, mid), we check the left side
            if is_even:
                r = mid - 2
            else:
                l = mid + 1
        else:
            # If nums[mid] is equal to neither neighbor, it's the single one
            return nums[mid]
    
    return nums[l]

```
