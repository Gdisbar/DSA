  - If str1[i] == 'F', the substring of word with size m starting at index i is not equal to str2, i.e., word[i..(i + m - 1)] != str2.

Return the lexicographically smallest possible string that can be generated by str1 and str2. If no string can be generated, return an empty string "".

## 1689. Partitioning Into Minimum Number Of Deci-Binary Numbers

A decimal number is called deci-binary if each of its digits is either 0 or 1 without any leading zeros. For example, 101 and 1100 are deci-binary, while 112 and 3001 are not.

Given a string n that represents a positive decimal integer, return the minimum number of positive deci-binary numbers needed so that they sum up to n.

```bash
Input: n = "32"
Output: 3
Explanation: 10 + 11 + 11 = 32
```

       

```python

"""
Take max deci-binary all 1 like "1111" -> "1110" -> .. -> until we have "0000" i.e we need to make all digits "0", so we try to make the max digit "0" and count 

Continues as long as there is at least one digit remaining to be reduced
 - In each "operation," you subtract 1 from every digit that is 
   greater than 0.
 - count how many times you perform this subtraction

Step 1: Subtract 1 from both -> "21" (total_ops = 1)
Step 2: Subtract 1 from both -> "10" (total_ops = 2)
Step 3: Subtract 1 from the remaining -> "00" (total_ops = 3)

The loop will run exactly as many times as the value of the largest 
digit. If the largest digit is 3, it takes 3 steps to reduce it to 0.

- Every other digit (which must be <= 3) will have reached 0 by 
 that time as well.

"""

def minPartitions(self, n: str) -> int:
    digits = [int(d) for d in n]
    # return max(digits) # this is the answer
    total_ops = 0
    # run till all digits are not 0
    while any(d > 0 for d in digits):
        # one ops - subtract 1 from every digit
        for i in range(len(digits)):
            if digits[i] > 0:
                digits[i] -=1
        total_ops +=1
    return total_ops
```

## 3776. Minimum Moves to Balance Circular Array



You are given a circular array balance of length n, where balance[i] is the net balance of person i.

In one move, a person can transfer exactly 1 unit of balance to either their left or right neighbor.

Return the minimum number of moves required so that every person has a non-negative balance. If it is impossible, return -1.

Note: You are guaranteed that at most 1 index has a negative balance initially.

 
```bash
Input: balance = [5,1,-4]
Output: 4
Explanation:

Move 1 unit from i = 1 to i = 2, resulting in balance = [5, 0, -3]
Move 1 unit from i = 0 to i = 2, resulting in balance = [4, 0, -2]
Move 1 unit from i = 0 to i = 2, resulting in balance = [3, 0, -1]
Move 1 unit from i = 0 to i = 2, resulting in balance = [2, 0, 0]
Thus, the minimum number of moves required is 4.
```

```python

"""
1. total balance < 0 : can't adjust
2. total balance >= 0 : no move required
3. find culpritIdx
    dist = 1
    while balance[culpritIdx] < 0:
        # if both same Idx - take only one
        total = leftIdxVal + rightIdxVal 
        # no extra move - take only required move
        tmp_balance = min(-balance[culpritIdx],total) 
        balance[culpritIdx] += tmp_balance
        move = tmp_balance * dist
        dist+=1

[_,2,  0,  0,-10,0,0,3,_,_]
   |-> |-> |->|
to move 1 unit from pos=1 to pos=4 , total moves/dist = 3
to move 2 unit = 2 * 3 = tmp_balance * dist
"""

def minMoves(self, balance: list[int]) -> int:
    n = len(balance)
    total_sum = sum(balance)
    
    # Find the last index that has a negative balance
    culprit_index = -1
    for i in range(n):
        if balance[i] < 0:
            culprit_index = i
    # If no negative values, no moves needed
    if culprit_index == -1:
        return 0
    # If the total sum is negative, it's impossible to balance
    if total_sum < 0:
        return -1
    
    moves = 0
    dist = 1
    
    # Continue until the culpritIndex is balanced
    while balance[culprit_index] < 0:
        right = (culprit_index + dist) % n
        left = (culprit_index - dist + n) % n
        # Sum available from neighbors
        available = balance[left] + balance[right]
        if left == right:
            available -= balance[right]

        # take only what is needed not more
        taken = min(-balance[culprit_index], available)
        moves += taken * dist
        balance[culprit_index] += taken
        dist += 1
        
    return moves
```

## 948. Bag of Tokens

You start with an initial power of power, an initial score of 0, and a bag of tokens given as an integer array tokens

Your goal is to maximize the total score by strategically playing these tokens. In one move, you can play an unplayed token in one of the two ways (but not both for the same token):

    Face-up: If your current power is at least tokens[i], you may play tokeni, losing tokens[i] power and gaining 1 score.

    Face-down: If your current score is at least 1, you may play tokeni, gaining tokens[i] power and losing 1 score.

Return the maximum possible score you can achieve after playing any number of tokens.
```bash

Input: tokens = [100], power = 50
Output: 0
Explanation: Since your score is 0 initially, you cannot play the token face-down. You also cannot play it face-up since your power (50) is less than tokens[0] (100).

Input: tokens = [200,100], power = 150
Output: 1
Explanation: Play token1 (100) face-up, reducing your power to 50 and increasing your score to 1.

There is no need to play token0, since you cannot play it face-up to add to your score. The maximum score achievable is 1.
```

```python
"""

1. sort array -> we decrease power from start & increase power from back
2. power>=tokens[i] -> incrase score,decrease power,increase i
3. power < tokens[i] -> we can increase power by decreasing score 'only 
 - if score >=1 and we have at least 2 elements on right to 
   increase score in future' otherwise we break out the loop
4. run loop till i <= j , when i==j it's the last token but i < j 
   skips it then no point of checking at least 2 elements exist or not

We can skip the window checking by maintaining a maxScore inside step 2
"""

def bagOfTokensScore(self, tokens: List[int], power: int) -> int:
    score = 0
    max_score = 0
    tokens = sorted(tokens)
    i = 0
    j = len(tokens)-1
    while i <= j:
        # incrase score,decrease power
        if power>=tokens[i]:
            power-=tokens[i]
            score+=1
            i+=1
            max_score = max(max_score,score)
        else:
            if score >= 1:
                score-=1
                power+=tokens[j]
                j-=1
            else:
                break
    
    return max_score
```

Return the resulting string. If there is no way to replace a character to make it not a palindrome, return an empty string.

For example, "abcc" is lexicographically smaller than "abcd"

```bash
Input: palindrome = "abccba"
Output: "aaccba"
Explanation: There are many ways to make "abccba" not a palindrome, such as "zbccba", "aaccba", and "abacba".
Of all the ways, "aaccba" is the lexicographically smallest.

Input: palindrome = "a"
Output: ""
Explanation: There is no way to replace a single character to make "a" not a palindrome, so return an empty string.
    
```


```python
"""
why check only n//2 chars ?
- If the string has an odd length (e.g., "aba"), the middle character (at index 1) is part of the symmetry. Changing the middle character of an odd-length palindrome doesn't actually stop it from being a palindrome.

all 'a' case or odd string length ?
- If the loop finishes and we haven't returned, it means the first half of the string consists entirely of 'a's. Example: "aaaaa".
- all n/2 exhausted : need to change last char n-1 to 'b' 

Example - "aba", "aabbaa", "aaaaa"
"aba" -> "abb" (change outside loop n-1)
"aabbaa" -> "aaabaa" (change inside loop idx=2)
"aaaaa" -> "aaaab" (change outside loop n-1)
"""
def breakPalindrome(self, palindrome: str) -> str:
    pal= list(palindrome)
    n = len(pal)
    if n==1: return ""
    # if n==even , this loop will return 
    for i in range(n//2):
        if pal[i]>'a':
            pal[i]='a'
            return "".join(pal)
    pal[n-1]='b'
    return "".join(pal)
```

```code
               0 1 2 3 4
colors =      "a b a a c"
               0 1 2 3 4
neededTime = [ 1,2,3,4,5]

i = 0
--------------
time = min(0,1)=0
prev = max(0,1)=1

i=1
---------
prev = 0 #not same
time = 0+min(0,2)=0+0=0
prev = max(0,2)=2

i=2
---------
prev = 0 #not same
time = 0+min(0,3)=0+0=0
prev = max(0,3)=3

i=3
---------
time = 0+min(3,4)=3
prev = max(3,4) = 4

i=4
---------
prev = 0 #not same
time = 3+min(0,5)=3+0=3
prev = max(0,5)=5


```
